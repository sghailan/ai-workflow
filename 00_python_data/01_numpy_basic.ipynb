# NumPy — Arrays y operaciones numéricas

## Creación de arrays

Recordamos que NumPy gira alrededor de una estructura clara: el array. Parecido a una lista de Python pero más potente

```python
a = np.array([1,2,3,4,5]) # array de una dimensión
print(a)
```

```python
print(type(a)) # tipo del objeto
```

ndarray significa N-dimensional array. Puede tener 1, 2, 3 o más dimensiones

```python
print(a.shape) # dimensión array
print(a.dtype) # tipo de datos en el objeto
```

(5,) significa 1 dimensión con 5 elementos. int64 es el tipo de dato: enteros de 64 bits.

```python
b = np.array([[1,2,3],[4,5,6]]) # array de dos dimensiones : matriz con 2 filas y 3 columnas. Pones dentro por filas
print(b)
```

```python
print(b.shape)
```

```python
print(b.ndim) # numero de dimensiones-> dos
print(b.size) # numero de datos en el array
```

Por ahora, hemos usado la funcion array() para crear arrays de una dimension como [1,2] o de dos dimensiones como [[1,2],[3,4]]. El atributo shape para ver número de filas y de columnas, ndim para ver el número de dimensiones del array y size para ver el número de datos dentro del array.

Vemos otras formas de crear los arrays, de manera no tan directa

```python
c = np.zeros((3,4)) # llenar el array de ceros
print(c)

print(np.ones((2,4))) # y de unos 
```

```python
print(np.arange(0,10,2)) # del 0 al 10 dame numeros de 2 en 2
```

```python
print(np.linspace(0,1,5)) # del 0 y 1 dame 5 numeros espaciados igual
```

linspace divide un intervalo en partes iguales.

np.linspace(0, 1, 5) significa: dame 5 números entre 0 y 1 espaciados igual.

Resultado: [0, 0.25, 0.5, 0.75, 1.0]

La diferencia con arange es que en arange tú controlas el paso, en linspace tú controlas cuántos números quieres.

```python
a_random = np.random.randint(0,100,size=(4,5)) # array random 

print(a_random)
```

## Indexing y Slicing

Indexing — acceder a un elemento concreto por su posición. Como decir "dame el elemento de la fila 0, columna 2".

Slicing — coger un trozo del array. Como decir "dame todas las filas de la columna 1" o "dame desde la columna 1 hasta el final".

```python
# se ha creado b = np.array([[1,2,3],[4,5,6]])
print(b)
```

```python
# accedemos a elementos del array de dos dimensiones
print(b[0]) # primera fila
print(b[1]) # segunda fila 
```

```python
print(b[0,2]) # fila 0 columna 2-> me espero 3
```

```python
print(b[:,1]) # todas las filas columna 1 (parecido a lo anterior pero con : para referirse a todas las filas)
```

```python
print(b[0,1:]) # la fila 0 desde la columna 1 a todo lo demas 
```

## Reshape

Reshape es cambiar la forma del array sin cambiar los datos. Por ejemplo, un array de 12 elementos lo puedes reorganizar como una matriz de 3x4, o de 2x6, o de 4x3 — siempre que el total de elementos sea el mismo.

```python
a = np.arange(1,13) # es como poner .arange(1,13,1)

print(a)
print(a.shape) # un vector 
```

```python
b = a.reshape(3,4)

print(b)
print(b.shape)
```

```python
c = a.reshape(2,6)

print(c)
print(c.shape)
```

```python
d= a.reshape(4,3)

print(d)
print(d.shape)
```

## Broadcasting

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([10, 20, 30])
print(a)
print(b)

print(a + b)
```

numpy "expande" automáticamente el array pequeño para operar con el grande, sin necesidad de duplicarlo manualmente.
Es muy útil para operaciones como normalizar datos, restar la media, etc.
