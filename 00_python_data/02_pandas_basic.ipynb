# Pandas Básico — Manipulación de datos tabulares

## 1. Series y DataFrames

Una Series es una columna — array 1D con etiquetas. Un DataFrame es una tabla — conjunto de Series.

```python
s = pd.Series([10,20,30,40,50])
print(s)
```

```python
print(type(s))
```

np.array([10, 20, 30]) — solo valores, índice numérico implícito

pd.Series([10, 20, 30]) — valores + índice explícito con etiquetas

```python
s = pd.Series([10,20,30], index = ['manzana', 'pera', 'naranja'])

print(s)
```

```python
print(s['manzana']) # accedemos ahora por nombre, no por posición como haciamos en los arrays
```

Un DataFrame es simplemente un diccionario de Series. Cada columna es una Series y todas comparten el mismo índice.

DataFrame
├── columna 'producto'  → Series de strings
├── columna 'precio'    → Series de floats
└── columna 'kilos'     → Series de ints

```python
datos = {
    'producto': ['manzana', 'pera', 'naranja', 'fresa'],
    'precio': [1.2,0.8,0.5,2.1],
    'kilos': [100,150,200,80]
}

df = pd.DataFrame(datos)

print(df) # diccionario de series de las que hemos visto
print(type(df))
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

Y por dentro cada Series es un array de numpy. Es decir:
dict → DataFrame → Series → numpy array

## 4. Indexing — loc e iloc

```python
df.head()
```

```python
print(df.loc[0])           # fila con índice 0 interno
```

```python
print(df.iloc[0])          # primera fila por posición q ves
```

```python
print(df.loc[0, 'nombre']) # fila 0, columna nombre
```

```python
print(df.iloc[0, 1])       # fila 0, columna 1
```

```python
print(df.loc[0:3, 'nombre':'ciudad'])   # filas 0 a 3, columnas nombre hasta ciudad
print(df.iloc[0:3, 0:3])                # filas 0 a 3, columnas 0 a 3
```

 Una cosa importante — loc es inclusivo en ambos extremos (incluye el 3), iloc es exclusivo al final (no incluye el 3), igual que el slicing de Python.

## 5. Filtrado

```python
# Filtrar empleados con salario mayor de 40000
print(df[df['salario'] > 40000])
```

```python
# Una condición
df[df['salario'] > 40000]

# Dos condiciones — AND
df[(df['salario'] > 40000) & (df['ciudad'] == 'Madrid')]

# Dos condiciones — OR
df[(df['salario'] > 40000) | (df['ciudad'] == 'Madrid')]

# isin — varios valores
df[df['ciudad'].isin(['Madrid', 'Barcelona'])]

# notna — filtrar nulos
df[df['salario'].notna()]
```

```python
# query — forma más limpia de filtrar
df.query("salario > 40000 and ciudad == 'Madrid'")

# ~ — negación
df[~df['ciudad'].isin(['Madrid', 'Barcelona'])]
```

## 6. Operaciones básicas

```python
# Añadir columna
df['salario_mensual'] = df['salario'] / 12

# Eliminar columna
df.drop(columns=['salario_mensual'], inplace=True)

# Renombrar columna
df.rename(columns={'nombre': 'name'}, inplace=True)

# Ordenar
df.sort_values('salario', ascending=False)

# Valores únicos
df['ciudad'].unique()
df['ciudad'].value_counts()

# Aplicar función
df['salario'].apply(lambda x: x * 1.1)

# Agrupar
df.groupby('ciudad')['salario'].mean()

# Nulos
df.isnull().sum()
df.dropna()
df.fillna(0)
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200

df = pd.DataFrame({
    'nombre': np.random.choice(['Ana', 'Luis', 'María', 'Carlos', 'Elena', 'Pedro', 'Laura', 'Juan', 'Sofía', 'David'], n),
    'edad': np.random.randint(18, 65, n),
    'ciudad': np.random.choice(['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Bilbao'], n),
    'salario': np.random.randint(18000, 80000, n),
    'departamento': np.random.choice(['Ventas', 'IT', 'RRHH', 'Marketing', 'Finanzas'], n),
    'años_empresa': np.random.randint(1, 25, n),
    'satisfaccion': np.random.choice([1, 2, 3, 4, 5], n),
    'remoto': np.random.choice([True, False], n),
    'formacion': np.random.choice(['Bachillerato', 'FP', 'Grado', 'Master', 'Doctorado'], n),
    'ventas_anuales': np.random.randint(0, 500000, n)
})

# Introducir algunos nulos
df.loc[np.random.choice(n, 20), 'salario'] = np.nan
df.loc[np.random.choice(n, 15), 'satisfaccion'] = np.nan

df.to_csv("../datasets/empleados_rico.csv", index=False)
print("Guardado!")
df.head()
```

### Ejercicio 3 — Operaciones
Crea una columna `salario_por_año` que divida el salario entre los años en la empresa. Ordena el DataFrame por esta columna de mayor a menor.

### Ejercicio 5 — Groupby avanzado
Agrupa por ciudad y departamento a la vez y calcula el salario máximo y mínimo de cada combinación.

### Ejercicio 6 — Apply
Crea una columna `nivel_seniority` basada en años_empresa: menos de 3 años → 'Junior', entre 3 y 10 → 'Mid', más de 10 → 'Senior'.

```python
df_ejers['nivel_seniority']= df_ejers['años_empresa'].apply(lambda x: 'Junior' if x<3 else ('Mid' if x<=10 else 'Senior'))
```

```python
df_ejers.head()
```

### Ejercicio 7 — Nulos
Rellena los nulos de salario con la media del salario de su mismo departamento (no la media global).

```python
df_ejers['salario'].isna().sum()
```

```python
df_ejers.groupby('departamento')['salario'].mean()
```

```python
df_ejers['salario'] = df_ejers['salario'].fillna(df_ejers.groupby('departamento')['salario'].transform('mean')) # rellenas los nulos con lo que corresponda de la media del transform
```

La clave es transform — en vez de devolver una media por grupo, devuelve el valor de la media para cada fila según su departamento

### Ejercicio 8 — Filtrado complejo
Encuentra los empleados cuyo salario está por encima de la media de su ciudad y además tienen satisfacción mayor o igual a 4.

```python
# 1 media de salarios por ciudad para cada registro con la funcion transform
media_ciudad = df_ejers.groupby('ciudad')['salario'].transform('mean')

df_ejers[(df_ejers['salario']> media_ciudad) & (df_ejers['satisfaccion']>=4)]
```

### Ejercicio 10 — Resumen final
Crea un DataFrame resumen con: número de empleados, salario medio, satisfacción media y total de ventas anuales por departamento, ordenado por salario medio descendente.

```python
df_ejers.groupby('departamento').agg(
    numero_empleados=('salario', 'count'),
    salario_medio=('salario', 'mean'),
    satisfaccion_media=('satisfaccion', 'mean'),
    total_ventas=('ventas_anuales', 'sum')
).sort_values('salario_medio', ascending=False)
```
